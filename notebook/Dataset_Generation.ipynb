```python
from PIL import Image, ImageDraw, ImageFont
import os
import re
import csv

# Define the width and height of the image
image_width = 1400
image_height = 100
start_index = 19472

# Path to the input text file and output directory
input_file_path = 'Khmer1.txt'
output_dir = 'new_dataset_moulight/images/Moulight'
label_file_path = 'new_dataset_moulight/labels/labels.csv'
# label_file_path = 'new_dataset/labels/labels.csv'

# Create the output directory if it doesn't exist
if not os.path.exists(output_dir):
    os.makedirs(output_dir)


# Path to the Khmer font file
font_path = 'Fonts/KHMEROSMUOLLIGHT.TTF'  # Ensure the path to the Khmer font file is correct

# Load the font
font_size = 13  # Adjusted font size for better visibility
font = ImageFont.truetype(font_path, font_size)

# Read the input text file
with open(input_file_path, 'r', encoding='utf-8') as file:
    lines = file.readlines()

# Function to remove all types of spaces from the text
def remove_all_spaces(text):
    # Use a regular expression to remove all types of whitespace characters
    return re.sub(r'\s+', '', text)

# Open the CSV file for writing
with open(label_file_path, 'w', encoding='utf-8', newline='') as csvfile:
    # Create a CSV writer object
    csvwriter = csv.writer(csvfile)

    # Write the header row
    csvwriter.writerow(['image', 'label'])
    
    # Convert each line of text to an image
    for i, line in enumerate(lines, start=start_index):
        # Remove all spaces (including non-breaking spaces) from the line
        cleaned_line = remove_all_spaces(line.strip())
        if len(cleaned_line) == 0:
            continue  # Skip empty lines

        # Create a new image with a white background
        image = Image.new('RGB', (image_width, image_height), color='white')
        draw = ImageDraw.Draw(image)

        # Calculate the position to draw the text
        text_bbox = draw.textbbox((0, 0), cleaned_line, font=font)
        text_width = text_bbox[2] - text_bbox[0]
        text_height = text_bbox[3] - text_bbox[1]
        text_x = (image_width - text_width) // 2
        text_y = (image_height - text_height) // 2

        # Draw the text onto the image
        draw.text((text_x, text_y), cleaned_line, font=font, fill='black')

        # Save the image
        output_path = os.path.join(output_dir, f'Khmer_txt_{i}.png')
        image.save(output_path)

        # Write the image filename and label to the CSV file
        csvwriter.writerow([f'Khmer_txt_{i}.png', cleaned_line])

print('Conversion complete. Images and labels saved in:', output_dir)
```
